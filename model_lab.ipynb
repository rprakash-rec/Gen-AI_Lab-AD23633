```python
# Create a dummy text document
text_content = """
Artificial intelligence (AI) is intelligence demonstrated by machines, unlike the natural intelligence displayed by humans and animals. Leading AI textbooks define the field as the study of "intelligent agents": any device that perceives its environment and takes actions that maximize its chance of successfully achieving its goals.

AI applications include advanced web search engines, recommendation systems (used by YouTube, Amazon, and Netflix), understanding human speech (such as Siri and Alexa), self-driving cars, and generative AI tools like ChatGPT and Google Bard.

Machine learning (ML) is a subset of AI that focuses on the development of algorithms that allow computers to learn from data without being explicitly programmed. Deep learning is a further subset of machine learning that uses neural networks with many layers (deep neural networks) to model complex patterns in data.

The history of AI dates back to ancient myths and philosophical discussions. The formal field of AI research was founded at a workshop at Dartmouth College in 1956. Since then, AI has experienced several cycles of optimism, disillusionment (known as 'AI winters'), and renewed interest.

Recent advancements, especially in computational power and large datasets, have led to significant breakthroughs, particularly in deep learning and natural language processing. This has propelled AI into many aspects of daily life and industry.
"""

with open("sample_document.txt", "w") as f:
    f.write(text_content)

print("Created 'sample_document.txt'")
```

```python
from langchain_community.document_loaders import TextLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter

# Load the document
loader = TextLoader("sample_document.txt")
documents = loader.load()

# Split the document into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,  # Max characters per chunk
    chunk_overlap=50 # Characters to overlap between chunks
)
chunks = text_splitter.split_documents(documents)

print(f"Original document has {len(documents)} page(s).")
print(f"Split into {len(chunks)} chunk(s).")
print("First chunk:\n", chunks[0].page_content)
```

```python
from langchain_community.embeddings import SentenceTransformerEmbeddings
from langchain_community.vectorstores import Chroma

# Initialize the embedding model
# This model will download if not already present.
embeddings_model = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")

# Create a vector store from the chunks
# This will embed the chunks and store them in ChromaDB
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings_model,
    persist_directory="./chroma_db" # Directory to save the vector store
)

print("Vector store created and persisted to './chroma_db'")
```

```python
from langchain_community.llms import Ollama

# Initialize Ollama LLM
# Ensure Ollama server is running locally and you have a model like 'llama2' pulled.
# For example, run `ollama pull llama2` in your terminal.
llm = Ollama(model="llama2") # You can change 'llama2' to other models like 'mistral', 'phi', etc.

print(f"LLM initialized with model: {llm.model}")
print("If this fails, ensure Ollama is running and the model is available.")
```

```python
from langchain.chains.combine_documents import create_stuff_documents_chain
from langchain.chains import create_retrieval_chain
from langchain_core.prompts import ChatPromptTemplate


retriever = vectorstore.as_retriever()

prompt = ChatPromptTemplate.from_template(
    """Answer the following question based only on the provided context:

{context}

Question: {input}"""
)


document_chain = create_stuff_documents_chain(llm, prompt)


rag_chain = create_retrieval_chain(retriever, document_chain)

print("RAG chain successfully built.")
```

```python

question1 = "What is artificial intelligence and what are some of its applications?"
response1 = rag_chain.invoke({"input": question1})
print(f"\nQuestion 1: {question1}")
print(f"Answer 1: {response1['answer']}")
: Informatio
question2 = "Explain machine learning and deep learning."
response2 = rag_chain.invoke({"input": question2})
print(f"\nQuestion 2: {question2}")
print(f"Answer 2: {response2['answer']}")


question3 = "When was the formal field of AI research founded and what factors led to recent breakthroughs?"
response3 = rag_chain.invoke({"input": question3})
print(f"\nQuestion 3: {question3}")
print(f"Answer 3: {response3['answer']}")


question4 = "Who invented the internet?"
response4 = rag_chain.invoke({"input": question4})
print(f"\nQuestion 4: {question4}")
print(f"Answer 4: {response4['answer']}")
```
